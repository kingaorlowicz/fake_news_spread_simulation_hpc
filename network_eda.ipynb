```python
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
```

```python
G = nx.read_edgelist(r"data\facebook_combined.txt", nodetype=int)

degrees = [deg for _, deg in G.degree()]

print(f"People (nodes): {G.number_of_nodes()}")
print(f"Connections (edges): {G.number_of_edges()}")
print(f"Average number of friends: {np.mean(degrees):.2f}")
print(f"Median number of friends: {np.median(degrees):.0f}")
print(f"Largest hub: {np.max(degrees)} connections")

# Top 5 hubs
top_hubs = sorted(G.degree(), key=lambda x: x[1], reverse=True)[:5]
print("\nTop 5 hubs (ID: number of friends):", top_hubs)

# Wykres
plt.figure(figsize=(10, 5))
plt.hist(degrees, bins=50, color="steelblue", edgecolor="black", alpha=0.8)
plt.title("Distribution of number of friends in Facebook network (SNAP)")
plt.xlabel("Number of friends (node degree)")
plt.ylabel("Number of people")
plt.yscale("log")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```
